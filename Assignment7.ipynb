```python
# Challenge 1
# Abstract Class & Abstract Method

from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car is starting...")


car = Car()
car.start()
```

```python
# Challenge 2
# Abstract Class with Multiple Child Classes

from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):

    def sound(self):
        print("Dog barks")


class Cat(Animal):

    def sound(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

```python
# Challenge 3
# Polymorphism Using Method Overriding

class Shape:

    def draw(self):
        print("Drawing Shape")


class Circle(Shape):

    def draw(self):
        print("Drawing Circle")


class Rectangle(Shape):

    def draw(self):
        print("Drawing Rectangle")


circle = Circle()
rectangle = Rectangle()

circle.draw()
rectangle.draw()
```

```python
# Challenge 4
# Polymorphism with Loop

class Bird:

    def fly(self):
        print("Bird is flying")


class Sparrow(Bird):

    def fly(self):
        print("Sparrow flies low")


class Eagle(Bird):

    def fly(self):
        print("Eagle flies high")


birds = [Sparrow(), Eagle()]

for bird in birds:
    bird.fly()
```

```python
# Challenge 5
# Abstract Class + Polymorphism

from abc import ABC, abstractmethod

class Employee(ABC):

    @abstractmethod
    def work(self):
        pass


class Developer(Employee):

    def work(self):
        print("Developer writes code")


class Designer(Employee):

    def work(self):
        print("Designer creates UI designs")


employees = [Developer(), Designer()]

for employee in employees:
    employee.work()
```
