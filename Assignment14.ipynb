Questions

Q1. What is Data Exploration?

Data Exploration is the process of understanding and analyzing a dataset before using it for Machine Learning.


    

Q2. What is a Feature in Machine Learning?

A Feature is an input variable used by a Machine Learning model to make predictions.

Example:

Age, Salary, Gender



Q3. What is a Target Variable?

A Target Variable is the output that a Machine Learning model tries to predict.

In this dataset:

Purchased

is the target variable.


    

Q4. What is Correlation?

Correlation shows the relationship between two variables. It tells us how strongly two variables are related.

The correlation value ranges from -1 to +1.

+1 → Strong positive relationship
0 → No linear relationship
-1 → Strong negative relationship

    
    
Q5. Why is Encoding required in Machine Learning?

Encoding is required to convert categorical/text data into numerical values so that Machine Learning algorithms can process the data.

Example:

Male → 1
Female → 0



Q6. What is the difference between Training Data and Testing Data?
Training Data: Used to train the Machine Learning model.
Testing Data: Used to test how well the trained model performs on unseen data.

For example, a dataset can be divided into 80% training data and 20% testing data.
