```python
# Question 4
# Create a BankAccount class with deposit() and withdraw() methods.

class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(amount, "deposited successfully.")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(amount, "withdrawn successfully.")
        else:
            print("Insufficient Balance.")

    def display_balance(self):
        print("Current Balance :", self.balance)

# Creating object
account = BankAccount(10000)

account.display_balance()
account.deposit(3000)
account.withdraw(2500)
account.display_balance()
```

```python
# Question 8
# Create a Product class with a class variable tax_rate and calculate the final product price.

class Product:

    tax_rate = 0.18

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def final_price(self):
        tax = self.price * Product.tax_rate
        total = self.price + tax

        print("Product :", self.name)
        print("Original Price :", self.price)
        print("Final Price :", total)

# Creating object
product = Product("Laptop", 50000)

# Display final price
product.final_price()
```

```python
# Question 11
# Create a Calculator class with static methods
# add(), subtract(), multiply(), and divide().

class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def subtract(a, b):
        return a - b

    @staticmethod
    def multiply(a, b):
        return a * b

    @staticmethod
    def divide(a, b):
        if b != 0:
            return a / b
        else:
            return "Division by zero is not possible."

print("Addition:", Calculator.add(20, 10))
print("Subtraction:", Calculator.subtract(20, 10))
print("Multiplication:", Calculator.multiply(20, 10))
print("Division:", Calculator.divide(20, 10))
```

```python
# Question 12
# Create a TemperatureConverter class with a static method
# to convert Celsius to Fahrenheit.

class TemperatureConverter:

    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (celsius * 9 / 5) + 32

celsius = 30
fahrenheit = TemperatureConverter.celsius_to_fahrenheit(celsius)

print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
```

```python
# Question 16
# Create a Shape class and inherit
# Circle and Rectangle classes from it.

class Shape:

    def display(self):
        print("This is a Shape")

class Circle(Shape):

    def area(self, radius):
        area = 3.14 * radius * radius
        print("Area of Circle:", area)

class Rectangle(Shape):

    def area(self, length, width):
        area = length * width
        print("Area of Rectangle:", area)

circle = Circle()
rectangle = Rectangle()

circle.display()
circle.area(7)

rectangle.display()
rectangle.area(10, 5)
```

```python
# Question 19
# Create a BankAccount class and implement
# encapsulation using a private balance variable.

class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient Balance")

    def get_balance(self):
        return self.__balance

account = BankAccount(10000)

print("Balance:", account.get_balance())

account.deposit(5000)
account.withdraw(3000)

print("Updated Balance:", account.get_balance())
```

```python
# Question 21
# Create a Library Management System using
# classes, objects, constructors, and methods.

class Library:

    def __init__(self, book_name, author):
        self.book_name = book_name
        self.author = author
        self.available = True

    def issue_book(self):
        if self.available:
            self.available = False
            print(self.book_name, "has been issued.")
        else:
            print(self.book_name, "is not available.")

    def return_book(self):
        self.available = True
        print(self.book_name, "has been returned.")

    def display(self):
        print("Book Name :", self.book_name)
        print("Author :", self.author)
        print("Available :", self.available)

# Creating object
book = Library("Python Programming", "John")

book.display()
book.issue_book()
book.display()
book.return_book()
book.display()
```

```python
# Question 22
# Create a Hospital Management System using
# inheritance and encapsulation concepts.

class Hospital:

    def __init__(self, hospital_name):
        self.hospital_name = hospital_name

class Patient(Hospital):

    def __init__(self, hospital_name, patient_name, disease):
        super().__init__(hospital_name)
        self.__patient_name = patient_name
        self.__disease = disease

    def display(self):
        print("Hospital :", self.hospital_name)
        print("Patient :", self.__patient_name)
        print("Disease :", self.__disease)

patient = Patient("City Hospital", "Reva", "Fever")

patient.display()
```

```python
# Question 23
# Create a School Management System using
# class variables, class methods, and instance methods.

class School:

    school_name = "ABC Public School"

    def __init__(self, student_name, standard):
        self.student_name = student_name
        self.standard = standard

    @classmethod
    def change_school(cls, new_name):
        cls.school_name = new_name

    def display(self):
        print("Student :", self.student_name)
        print("Class :", self.standard)
        print("School :", School.school_name)

student = School("Reva", "SY AIML")

print("Before Updating School Name")
student.display()

School.change_school("XYZ International School")

print("\nAfter Updating School Name")
student.display()
```

```python
# Question 25
# Create a Mini ATM System using
# encapsulation, constructors, and instance methods.

class ATM:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print(amount, "deposited successfully.")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(amount, "withdrawn successfully.")
        else:
            print("Insufficient Balance.")

    def check_balance(self):
        print("Current Balance :", self.__balance)

# Creating object
atm = ATM(10000)

atm.check_balance()
atm.deposit(5000)
atm.withdraw(3000)
atm.check_balance()
```
