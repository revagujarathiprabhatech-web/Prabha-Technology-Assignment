```python
# Pandas Training - Day 2 Assignment
# Topic: Data Selection & Filtering

import pandas as pd

# Create Employee DataFrame
data = {
    "Employee ID": [101, 102, 103, 104, 105],
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Rohan"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales"],
    "Salary": [45000, 35000, 55000, 40000, 60000],
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Pune"]
}

df = pd.DataFrame(data)

# Display DataFrame
df
```

```python
# 1. Display first 3 rows

df.head(3)
```

```python
# 2. Display last 2 rows

df.tail(2)
```

```python
# 3. Display dataset shape

df.shape
```

```python
# 4. Display column names

df.columns
```

```python
# 5. Display employees whose salary is greater than 40000

df[df["Salary"] > 40000]
```

```python
# 6. Display employees from Pune

df[df["City"] == "Pune"]
```
