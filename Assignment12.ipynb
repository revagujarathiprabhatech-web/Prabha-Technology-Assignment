Data Cleaning

```python
# Pandas Training - Day 3 Assignment
# Topic: Data Cleaning

import pandas as pd

# Create Student DataFrame with missing values and one duplicate record

data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Priya"],
    "Age": [20, None, 21, None, None],
    "Marks": [85, 90, None, 78, 90]
}

df = pd.DataFrame(data)

# Display DataFrame
df
```

```python
# Find missing values

df.isnull()
```

```python
# Count missing values in each column

df.isnull().sum()
```

```python
# Replace all missing values with 0

df_zero = df.fillna(0)

df_zero
```

```python
# Calculate average age

average_age = df["Age"].mean()

# Replace missing age with average age

df["Age"] = df["Age"].fillna(average_age)

df
```

```python
# Identify duplicate records

df.duplicated()
```

```python
# Remove duplicate records

df = df.drop_duplicates()

df
```
