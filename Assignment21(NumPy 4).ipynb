```python
# Question 25

salary = np.array([
    [25000, 30000, 35000],
    [40000, 45000, 50000]
])

# 1. Original Matrix
print("1. Original Salary Matrix:")
print(salary)

# 2. Reshape into 3 × 2
reshaped = salary.reshape(3, 2)

print("\n2. Reshaped 3 × 2 Matrix:")
print(reshaped)

# 3. Flatten the reshaped matrix
flattened = reshaped.flatten()

print("\n3. Flattened Array:")
print(flattened)

# 4. Transpose of original matrix
transposed = salary.T

print("\n4. Transpose:")
print(transposed)

# 5. Increase every salary by ₹2000
updated_salary = salary + 2000

print("\n5. Salary After ₹2000 Increase:")
print(updated_salary)

# 6. Create a copy and modify one value
salary_copy = salary.copy()
salary_copy[0, 0] = 99999

print("\n6. Original Matrix After Copy Modification:")
print(salary)

print("Copy After Modification:")
print(salary_copy)

# 7. Create a view and modify one value
salary_view = salary.view()
salary_view[0, 1] = 88888

print("\n7. Original Matrix After View Modification:")
print(salary)

print("View After Modification:")
print(salary_view)

# 8. Compare outputs
print("\n8. Comparison:")
print("Copy is independent from original.")
print("View shares data with original.")
```
