```python
# Q6

student = {
    "name": "Reva",
    "age": 19,
    "city": "Pune"
}

print("Keys:")
for key in student.keys():
    print(key)

print("\nValues:")
for value in student.values():
    print(value)

print("\nKey-Value Pairs:")
for key, value in student.items():
    print(key, ":", value)
```

Section 3 – Functions

```python
# Q21

def welcome():
    print("Welcome to Python Programming")

welcome()
```

```python
# Q22

def add(a, b):
    print("Sum =", a + b)

add(10,20)
```

```python
# Q23

def greet(name):
    print("Hello,", name)

greet("Rohit")
```

```python
# Q24

def square(num):
    print("Square =", num ** 2)

square(6)
```

```python
# Q25

def even_odd(num):
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")

even_odd(7)
```

```python
# Q26

def area(length, width):
    print("Area =", length * width)

area(10,5)
```

```python
# Q27

def greater(a,b):
    return max(a,b)

print(greater(15,20))
```

```python
# Q28

def numbers():
    for i in range(1,11):
        print(i)

numbers()
```

```python
# Q29

def total(lst):
    return sum(lst)

print(total([10,20,30,40]))
```

```python
# Q30

def result(marks):
    if marks >= 35:
        print("Pass")
    else:
        print("Fail")

result(45)
```

```python
# Q31

def details(name, city="Pune"):
    print("Name:", name)
    print("City:", city)

details("Reva")
```

```python
# Q32

def student(name, age):
    print("Name:", name)
    print("Age:", age)

student(age=19, name="Reva")
```

```python
# Q33

def numbers(*args):
    for num in args:
        print(num)

numbers(10,20,30,40,50)
```

```python
# Q34

def max_min(lst):
    return max(lst), min(lst)

maximum, minimum = max_min([12,45,67,8,23])

print("Maximum:", maximum)
print("Minimum:", minimum)
```

```python
# Q35

def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for letter in text:
        if letter in vowels:
            count += 1

    return count

print("Total Vowels:", count_vowels("Python Programming"))
```
