Sorting & Grouping

```python
# Pandas Training - Day 4 Assignment
# Topics: Sorting & Grouping

import pandas as pd

# Create Employee DataFrame

data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Rohan", "Neha"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "Salary": [45000, 35000, 55000, 40000, 50000, 60000]
}

df = pd.DataFrame(data)

# Display DataFrame
df
```

```python
# Sort salary in ascending order

df.sort_values("Salary", ascending=True)
```

```python
# Sort salary in descending order

df.sort_values("Salary", ascending=False)
```

```python
# Display unique departments

df["Department"].unique()
```

```python
# Count employees in each department

df["Department"].value_counts()
```

```python
# Find average salary of each department

df.groupby("Department")["Salary"].mean()
```

```python
# Find maximum salary of each department

df.groupby("Department")["Salary"].max()
```
