```python
# Q1. Create a DataFrame containing 5 students

import pandas as pd

data = {
    "Name": ["Aarav", "Riya", "Sneha", "Aditya", "Priya"],
    "Age": [20, 21, 19, 22, 20],
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai"],
    "Marks": [85, 92, 78, 67, 88]
}

df = pd.DataFrame(data)

# Save the DataFrame as students.csv
df.to_csv("students.csv", index=False)

print("DataFrame created successfully!")
```
