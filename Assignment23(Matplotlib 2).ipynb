```python
import matplotlib.pyplot as plt
```

```python
# Q1. Create a Vertical Bar Chart showing the sales of four products.

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [500, 800, 400, 300]

plt.bar(products, sales)
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
# Q2. Create a Horizontal Bar Chart showing employee salaries of five employees.

employees = ["Amit", "Riya", "Rahul", "Sneha", "Karan"]
salaries = [30000, 40000, 35000, 45000, 50000]

plt.barh(employees, salaries)
plt.title("Employee Salaries")
plt.xlabel("Salary")
plt.ylabel("Employees")
plt.show()
```

```python
# Q3. Create a Bar Chart showing the marks of six students.
# Requirements: Green bars, Black border, Grid.

students = ["A", "B", "C", "D", "E", "F"]
marks = [75, 85, 65, 90, 70, 80]

plt.bar(students, marks, color="green", edgecolor="black")
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid(axis="y")
plt.show()
```

```python
# Q4. Create a Product Sales Bar Chart using different colors for each bar.

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [500, 800, 400, 300]
colors = ["red", "blue", "green", "orange"]

plt.bar(products, sales, color=colors)
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
# Q5. Create a Bar Chart showing monthly expenses.
# Requirements: Figure Size (8,5), Orange bars, Horizontal Grid.

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
expenses = [12000, 15000, 10000, 18000, 14000, 16000]

plt.figure(figsize=(8, 5))
plt.bar(months, expenses, color="orange")
plt.title("Monthly Expenses")
plt.xlabel("Months")
plt.ylabel("Expenses")
plt.grid(axis="y")
plt.show()

```

```python
# Q6. Create a Pie Chart showing the percentage of payment methods.

methods = ["Cash", "UPI", "Card", "Net Banking"]
payments = [20, 50, 20, 10]

plt.pie(payments, labels=methods, autopct="%1.1f%%")
plt.title("Payment Methods")
plt.show()
```

```python
# Q7. Create a Pie Chart for monthly expenses.
# Requirements: Labels, Percentages (autopct), Shadow.

months = ["Jan", "Feb", "Mar", "Apr", "May"]
expenses = [10000, 15000, 12000, 18000, 14000]

plt.pie(expenses, labels=months, autopct="%1.1f%%", shadow=True)
plt.title("Monthly Expenses")
plt.show()
```

```python
# Q25. Final Practical Challenge
# Create a Sales Analysis Dashboard containing:
# Product Sales Bar Chart
# Payment Method Pie Chart
# Student Marks Distribution Histogram
# Study Hours vs Marks Scatter Plot
# Requirements: Use plt.subplot(), plt.suptitle(), titles,
# labels, grid on Bar Chart, Histogram and Scatter Plot,
# Figure Size (12,8).

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [500, 800, 400, 300]

methods = ["Cash", "UPI", "Card", "Net Banking"]
payments = [20, 50, 20, 10]

marks = [45, 55, 60, 65, 70, 72, 75, 78, 80, 82,
         85, 88, 90, 92, 95]

hours = [1, 2, 3, 4, 5, 6, 7, 8]
student_marks = [45, 50, 55, 65, 70, 75, 85, 90]

plt.figure(figsize=(12, 8))

# Bar Chart
plt.subplot(2, 2, 1)
plt.bar(products, sales, color="green", edgecolor="black")
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.grid(axis="y")

# Pie Chart
plt.subplot(2, 2, 2)
plt.pie(payments, labels=methods, autopct="%1.1f%%")
plt.title("Payment Methods")

# Histogram
plt.subplot(2, 2, 3)
plt.hist(marks, bins=5, color="purple", edgecolor="black")
plt.title("Student Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid()

# Scatter Plot
plt.subplot(2, 2, 4)
plt.scatter(hours, student_marks, color="red", marker="o", s=100)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()

plt.suptitle("Sales Analysis Dashboard")
plt.tight_layout()
plt.show()
```
