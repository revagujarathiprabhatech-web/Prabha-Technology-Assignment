```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder
```

```python
# Question 1: Find Missing Values

data = {
    "Name": ["Amit", "Rahul", "Sneha", "Priya"],
    "Age": [25, np.nan, 28, 30],
    "Salary": [30000, 45000, np.nan, 55000]
}

df = pd.DataFrame(data)

# Display dataset
print("Dataset:")
print(df)

# Display all missing values
print("\nMissing Values using isnull():")
print(df.isnull())

# Display total missing values in each column
print("\nTotal Missing Values:")
print(df.isnull().sum())
```

```python
# Question 2: Remove Missing Values

# Remove rows containing missing values
df_cleaned = df.dropna()

# Display updated dataset
print("Dataset after removing missing values:")
print(df_cleaned)
```

```python
# Question 3: Replace Missing Values Using Mean

data = {
    "Age": [20, 25, np.nan, 35, 40]
}

df_age = pd.DataFrame(data)

print("Original Dataset:")
print(df_age)

# Calculate mean
mean_age = df_age["Age"].mean()

# Replace missing value with mean
df_age["Age"] = df_age["Age"].fillna(mean_age)

print("\nDataset after replacing missing value with Mean:")
print(df_age)
```

```python
# Question 4: Replace Missing Values Using Median

data = {
    "Salary": [30000, 35000, 40000, np.nan, 90000]
}

df_salary = pd.DataFrame(data)

print("Original Dataset:")
print(df_salary)

# Calculate median
median_salary = df_salary["Salary"].median()

# Replace missing value with median
df_salary["Salary"] = df_salary["Salary"].fillna(median_salary)

print("\nDataset after replacing missing value with Median:")
print(df_salary)
```

```python
# Question 5: Replace Missing Values Using Mode

data = {
    "City": ["Pune", "Mumbai", "Pune", np.nan, "Delhi", "Pune"]
}

df_city = pd.DataFrame(data)

print("Original Dataset:")
print(df_city)

# Calculate mode
mode_city = df_city["City"].mode()[0]

# Replace missing value with mode
df_city["City"] = df_city["City"].fillna(mode_city)

print("\nDataset after replacing missing value with Mode:")
print(df_city)
```

```python
# Question 6: Use SimpleImputer

data = {
    "Age": [22, np.nan, 30, 35],
    "Salary": [25000, 35000, np.nan, 50000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Create SimpleImputer with Mean strategy
imputer = SimpleImputer(strategy="mean")

# Apply imputer
df[["Age", "Salary"]] = imputer.fit_transform(df[["Age", "Salary"]])

print("\nDataset after SimpleImputer:")
print(df)
```

```python
# Question 7: Apply Label Encoding

data = {
    "Gender": ["Male", "Female", "Male", "Female", "Male"]
}

df_gender = pd.DataFrame(data)

print("Original Dataset:")
print(df_gender)

# Create LabelEncoder
encoder = LabelEncoder()

# Apply Label Encoding
df_gender["Gender"] = encoder.fit_transform(df_gender["Gender"])

print("\nDataset after Label Encoding:")
print(df_gender)
```

```python
# Question 8: Apply One-Hot Encoding

data = {
    "City": ["Pune", "Mumbai", "Delhi", "Pune", "Mumbai"]
}

df_city = pd.DataFrame(data)

print("Original Dataset:")
print(df_city)

# Apply One-Hot Encoding
df_encoded = pd.get_dummies(df_city, columns=["City"], dtype=int)

print("\nDataset after One-Hot Encoding:")
print(df_encoded)
```

```python
# Question 9: Complete Data Preprocessing

data = {
    "Name": ["Amit", "Rahul", "Sneha", "Priya"],
    "Age": [25, np.nan, 28, 30],
    "Salary": [30000, 45000, np.nan, 55000],
    "Gender": ["Male", "Male", "Female", "Female"],
    "City": ["Pune", "Mumbai", "Delhi", "Pune"]
}

df = pd.DataFrame(data)

# Step 1: Display dataset
print("Step 1: Original Dataset")
print(df)

# Step 2: Find missing values
print("\nStep 2: Missing Values")
print(df.isnull().sum())

# Step 3: Replace numerical missing values using SimpleImputer
imputer = SimpleImputer(strategy="mean")

df[["Age", "Salary"]] = imputer.fit_transform(
    df[["Age", "Salary"]]
)

# Step 4: Apply Label Encoding to Gender
encoder = LabelEncoder()

df["Gender"] = encoder.fit_transform(df["Gender"])

# Step 5: Apply One-Hot Encoding to City
df = pd.get_dummies(df, columns=["City"], dtype=int)

# Step 6: Display final processed dataset
print("\nStep 6: Final Processed Dataset")
print(df)

# Verify missing values
print("\nMissing Values after Preprocessing:")
print(df.isnull().sum())
```

```python
# Question 10: Employee Data Preprocessing Challenge

data = {
    "Name": [
        "Amit", "Rahul", "Sneha", "Priya",
        "Rohan", "Neha", "Vikas", "Pooja"
    ],
    
    "Age": [
        25, np.nan, 28, 30,
        35, 27, np.nan, 32
    ],
    
    "Salary": [
        30000, 45000, np.nan, 55000,
        60000, 40000, 50000, np.nan
    ],
    
    "Experience": [
        2, 4, 3, 5,
        np.nan, 2, 6, 4
    ],
    
    "Gender": [
        "Male", "Male", "Female", "Female",
        "Male", "Female", "Male", "Female"
    ],
    
    "Department": [
        "IT", "HR", "Finance", "IT",
        "HR", "Finance", "IT", "HR"
    ]
}

df = pd.DataFrame(data)

# Step 1: Display the dataset
print("Step 1: Employee Dataset")
print(df)

# Step 2: Find missing values
print("\nStep 2: Missing Values")
print(df.isnull().sum())

# Step 3: Replace numerical missing values using SimpleImputer
imputer = SimpleImputer(strategy="mean")

df[["Age", "Salary", "Experience"]] = imputer.fit_transform(
    df[["Age", "Salary", "Experience"]]
)

# Step 4: Apply Label Encoding to Gender
encoder = LabelEncoder()

df["Gender"] = encoder.fit_transform(df["Gender"])

# Step 5: Apply One-Hot Encoding to Department
df = pd.get_dummies(
    df,
    columns=["Department"],
    dtype=int
)

# Step 6: Verify that there are no missing values
print("\nStep 6: Missing Values after Preprocessing")
print(df.isnull().sum())

# Step 7: Display final processed dataset
print("\nFinal Processed Dataset:")
print(df)
```
