```python
# Question 1
# Create your first line chart using the following data:
# X-axis: [1, 2, 3, 4, 5]
# Y-axis: [10, 20, 15, 25, 30]

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.plot(x, y)
plt.show()
```

```python
# Question 2
# Create a student marks line chart for five subjects.
# Marks: [72, 80, 76, 90, 95]
# Add Chart Title, X-axis Label, and Y-axis Label.

import matplotlib.pyplot as plt

subjects = ["Maths", "Science", "English", "Computer", "AI"]
marks = [72, 80, 76, 90, 95]

plt.plot(subjects, marks)

plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
```

```python
# Question 3
# Create an employee salary line chart using:
# [25000, 32000, 40000, 38000, 45000]
# Requirements: Blue Line, Circle (o) Marker, Grid.

import matplotlib.pyplot as plt

employees = [1, 2, 3, 4, 5]
salary = [25000, 32000, 40000, 38000, 45000]

plt.plot(employees, salary, color="blue", marker="o")

plt.title("Employee Salary")
plt.xlabel("Employee")
plt.ylabel("Salary")

plt.grid()
plt.show()
```

```python
# Question 4
# Plot the monthly sales for six months.
# Sales: [25000, 30000, 35000, 40000, 45000, 50000]
# Requirements: Green Line, Dashed (--) Line Style, Line Width = 3.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [25000, 30000, 35000, 40000, 45000, 50000]

plt.plot(months, sales, color="green", linestyle="--", linewidth=3)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
# Question 5
# Create a line chart for the following temperature data:
# [30, 32, 31, 35, 36, 34, 33]
# Requirements: Star (*) Marker, Red Line, Chart Title.

import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
temperature = [30, 32, 31, 35, 36, 34, 33]

plt.plot(days, temperature, color="red", marker="*")

plt.title("Temperature Data")
plt.xlabel("Day")
plt.ylabel("Temperature")

plt.show()
```

```python
# Question 6
# Compare the marks of Student A and Student B on the same graph.
# Student A: [75, 80, 78, 90, 88]
# Student B: [70, 85, 80, 86, 92]
# Requirements: Different Colors, Legend, Grid.

import matplotlib.pyplot as plt

subjects = ["Maths", "Science", "English", "Computer", "AI"]

student_a = [75, 80, 78, 90, 88]
student_b = [70, 85, 80, 86, 92]

plt.plot(subjects, student_a, color="blue", label="Student A")
plt.plot(subjects, student_b, color="red", label="Student B")

plt.title("Student Marks Comparison")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.legend()
plt.grid()

plt.show()
```

```python
# Question 7
# Create an attendance chart for five days.
# Attendance: [40, 42, 39, 45, 44]
# Requirements: Figure Size (8,5), Purple Line, Diamond (D) Marker,
# X-axis Label, Y-axis Label.

import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
attendance = [40, 42, 39, 45, 44]

plt.figure(figsize=(8, 5))

plt.plot(days, attendance, color="purple", marker="D")

plt.title("Attendance Chart")
plt.xlabel("Days")
plt.ylabel("Attendance")

plt.show()
```

```python
# Question 8
# Create a line chart showing product sales.
# Sales Data: [120, 180, 160, 220, 250]
# Requirements: Orange Line, Dotted (:) Line Style,
# Circle (o) Marker, Grid.

import matplotlib.pyplot as plt

products = ["Product 1", "Product 2", "Product 3", "Product 4", "Product 5"]
sales = [120, 180, 160, 220, 250]

plt.plot(products, sales, color="orange", linestyle=":", marker="o")

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.grid()
plt.show()
```

```python
# Question 9
# Create a line chart for website visitors over seven days.
# Visitors: [150, 180, 170, 200, 230, 210, 250]
# Requirements: Figure Size (10,5), Black Line, Plus (+) Marker,
# Title, Legend, Grid.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
visitors = [150, 180, 170, 200, 230, 210, 250]

plt.figure(figsize=(10, 5))

plt.plot(
    days,
    visitors,
    color="black",
    marker="+",
    label="Website Visitors"
)

plt.title("Website Visitors")
plt.xlabel("Days")
plt.ylabel("Visitors")

plt.legend()
plt.grid()

plt.show()
```

```python
# Question 10
# Create a professional Monthly Sales Report.
# Months: ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
# Sales: [25000, 30000, 28000, 40000, 45000, 50000]
# Requirements:
# Figure Size (10,5)
# Green Line
# Circle (o) Marker
# Dashed (--) Line Style
# Line Width = 3
# Chart Title
# X-axis Label
# Y-axis Label
# Grid
# Legend

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [25000, 30000, 28000, 40000, 45000, 50000]

plt.figure(figsize=(10, 5))

plt.plot(
    months,
    sales,
    color="green",
    marker="o",
    linestyle="--",
    linewidth=3,
    label="Monthly Sales"
)

plt.title("Monthly Sales Report")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid()
plt.legend()

plt.show()
```
