```python
# Machine Learning Data Preprocessing Challenge

# Import required libraries
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split


# ============================================================
# 1. LOAD DATASET
# ============================================================

# Read the CSV file
df = pd.read_csv("employee_dataset.csv")

# Display first 5 records
print("First 5 Records:")
print(df.head())


# Display dataset information
print("\nDataset Information:")
df.info()


# Display summary statistics
print("\nSummary Statistics:")
print(df.describe())


# ============================================================
# 2. HANDLE MISSING VALUES
# ============================================================

# Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())


# Fill missing Age values with median
df["Age"] = df["Age"].fillna(df["Age"].median())


# Verify missing values after filling
print("\nMissing Values After Filling:")
print(df.isnull().sum())


# ============================================================
# 3. REMOVE DUPLICATE RECORDS
# ============================================================

# Check number of duplicate rows
print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())


# Remove duplicate records
df = df.drop_duplicates()


# Verify duplicates have been removed
print("\nDuplicate Rows After Removal:")
print(df.duplicated().sum())


# ============================================================
# 4. ENCODE CATEGORICAL DATA
# ============================================================

# Identify categorical columns
categorical_columns = df.select_dtypes(include=["object"]).columns

print("\nCategorical Columns:")
print(categorical_columns)


# Encode Gender using LabelEncoder
label_encoder = LabelEncoder()

df["Gender"] = label_encoder.fit_transform(df["Gender"])


# Display encoded data
print("\nData After Encoding:")
print(df.head())


# ============================================================
# 5. FEATURE ENGINEERING
# ============================================================

# Create Experience_Level feature
df["Experience_Level"] = pd.cut(
    df["Experience"],
    bins=[-1, 5, 10, 15, 20],
    labels=[0, 1, 2, 3]
)

# Convert Experience_Level to integer
df["Experience_Level"] = df["Experience_Level"].astype(int)


# Display new feature
print("\nData After Feature Engineering:")
print(df.head())


# ============================================================
# 6. FEATURE SELECTION
# ============================================================

# Remove Name because it is not useful for prediction
df = df.drop("Name", axis=1)


# Display final selected features
print("\nFinal Selected Features:")
print(df.columns)


# ============================================================
# 7. CREATE FEATURES (X) AND TARGET (y)
# ============================================================

# Create input features
X = df.drop("Salary", axis=1)

# Create target variable
y = df["Salary"]


print("\nFeatures (X):")
print(X.head())

print("\nTarget (y):")
print(y.head())


# ============================================================
# 8. FEATURE SCALING
# ============================================================

# Create StandardScaler
scaler = StandardScaler()

# Scale all numerical feature columns
X_scaled = scaler.fit_transform(X)

# Convert scaled features back to DataFrame
X = pd.DataFrame(
    X_scaled,
    columns=X.columns,
    index=X.index
)


print("\nScaled Features:")
print(X.head())


# ============================================================
# 9. TRAIN-TEST SPLIT
# ============================================================

# Split data into 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


# Display shapes
print("\nTraining and Testing Shapes:")

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
```
