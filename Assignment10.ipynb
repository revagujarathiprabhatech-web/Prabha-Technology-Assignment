Introduction to Pandas, Series, DataFrame, CSV Files

```python
import pandas as pd

# 1. Create DataFrame
data = {
    "Name": ["Reva", "Aarav", "Priya", "Rahul", "Sneha"],
    "Age": [18, 19, 18, 20, 19],
    "City": ["Pune", "Mumbai", "Nashik", "Pune", "Mumbai"],
    "Marks": [85, 78, 92, 74, 88]
}

df = pd.DataFrame(data)

# 2. Display the DataFrame
print("Student Data:")
print(df)

# 3. Save the DataFrame into CSV file
df.to_csv("students.csv", index=False)

print("\nData saved successfully to students.csv")

# 4. Read the CSV file again
loaded_df = pd.read_csv("students.csv")

# 5. Display the loaded data
print("\nLoaded Data:")
print(loaded_df)
```
