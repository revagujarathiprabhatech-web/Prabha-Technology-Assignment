```python
# Question 1
import numpy as np

array1 = np.array([10, 20, 30, 40, 50])
array2 = np.array([5, 10, 15, 20, 25])

result = array1 + array2

print("Addition:", result)
```

```python
# Question 2
result = array1 - array2

print("Subtraction:", result)
```

```python
# Question 3
result = array1 * array2

print("Multiplication:", result)
```

```python
# Question 4
result = array1 / array2

print("Division:", result)
```
