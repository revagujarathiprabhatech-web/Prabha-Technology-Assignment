```python
# Question 25

matrix = np.arange(1, 26).reshape(5, 5)

print("Matrix:")
print(matrix)

print("\nFirst row:")
print(matrix[0])

print("\nLast row:")
print(matrix[-1])

print("\nFirst column:")
print(matrix[:, 0])

print("\nLast column:")
print(matrix[:, -1])

print("\nMiddle row:")
print(matrix[2])

print("\nMiddle column:")
print(matrix[:, 2])

print("\nTop-left 2 × 2 matrix:")
print(matrix[:2, :2])

print("\nBottom-right 2 × 2 matrix:")
print(matrix[-2:, -2:])

print("\nAll even numbers:")
print(matrix[matrix % 2 == 0])

print("\nAll odd numbers:")
print(matrix[matrix % 2 != 0])

print("\nNumbers greater than 15:")
print(matrix[matrix > 15])

print("\nNumbers less than 10:")
print(matrix[matrix < 10])

print("\nFancy Indexing - 1, 7, 13, 19, 25:")
indices = np.array([0, 6, 12, 18, 24])
print(matrix.flatten()[indices])

print("\nNegative Indexing - 25, 24, 23:")
print(matrix.flatten()[-1])
print(matrix.flatten()[-2])
print(matrix.flatten()[-3])

print("\nLast three rows:")
print(matrix[-3:])

print("\nFirst three columns:")
print(matrix[:, :3])
```
