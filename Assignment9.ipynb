File Handling

```python
# Challenge 1
# Create a file named student.txt and write your name into it.

file = open("student.txt", "w")

file.write("Reva Gujarathi")

file.close()

print("Name written successfully.")
```

```python
# Challenge 2
# Read data from student.txt.

file = open("student.txt", "r")

data = file.read()

print(data)

file.close()
```

```python
# Challenge 3
# Append your city name to the file.

file = open("student.txt", "a")

file.write("\nPune")

file.close()

print("City added successfully.")
```

```python
# Challenge 4
# Read the file using readline().

file = open("student.txt", "r")

print(file.readline())
print(file.readline())

file.close()
```

```python
# Challenge 5
# Check whether a file exists or not.

import os

if os.path.exists("student.txt"):
    print("File exists.")
else:
    print("File does not exist.")
```

```python
# Challenge 6
# Store 5 student names in a file and display them.

file = open("students.txt", "w")

for i in range(5):
    name = input(f"Enter student {i+1} name: ")
    file.write(name + "\n")

file.close()

print("\nStudent Names:")

file = open("students.txt", "r")

print(file.read())

file.close()
```
