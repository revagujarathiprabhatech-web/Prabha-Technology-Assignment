```python
# Q6. Create a 3 x 2 subplot layout and display any six
# different charts learned in Matplotlib.

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 12))

# 1. Line Chart
plt.subplot(3, 2, 1)
plt.plot([1, 2, 3, 4, 5], [10, 20, 15, 25, 30], marker="o")
plt.title("Line Chart")
plt.grid()

# 2. Bar Chart
plt.subplot(3, 2, 2)
plt.bar(["A", "B", "C", "D"], [20, 30, 25, 40])
plt.title("Bar Chart")
plt.grid(axis="y")

# 3. Pie Chart
plt.subplot(3, 2, 3)
plt.pie([30, 40, 20, 10], labels=["A", "B", "C", "D"],
        autopct="%1.1f%%")
plt.title("Pie Chart")

# 4. Histogram
plt.subplot(3, 2, 4)
plt.hist([45, 50, 55, 60, 65, 70, 75, 80], bins=4,
         edgecolor="black")
plt.title("Histogram")
plt.grid()

# 5. Scatter Plot
plt.subplot(3, 2, 5)
plt.scatter([1, 2, 3, 4, 5], [20, 30, 25, 40, 45])
plt.title("Scatter Plot")
plt.grid()

# 6. Horizontal Bar Chart
plt.subplot(3, 2, 6)
plt.barh(["A", "B", "C", "D"], [10, 20, 15, 25])
plt.title("Horizontal Bar Chart")
plt.grid(axis="x")

plt.tight_layout()
plt.show()
```

```python
# Q22. Read a CSV file and create:
# Pie Chart and Histogram.
# Display both in one figure.

import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
df = pd.read_csv("payment_sales.csv")

plt.figure(figsize=(10, 5))

# Pie Chart
plt.subplot(1, 2, 1)

payment_data = df.groupby("Payment_Method")["Payment_Amount"].sum()

plt.pie(payment_data.values,
        labels=payment_data.index,
        autopct="%1.1f%%")

plt.title("Payment Method Distribution")

# Histogram
plt.subplot(1, 2, 2)

plt.hist(df["Rating"], bins=5, edgecolor="black")

plt.title("Customer Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Customers")
plt.grid()

plt.tight_layout()
plt.show()
```

```python
# Q23. Read a CSV file and create a dashboard containing:
# Line Chart, Bar Chart, Scatter Plot and Pie Chart.
# Add Figure Size, Main Title, Grid and tight_layout().

import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
df = pd.read_csv("sales_dashboard.csv")

plt.figure(figsize=(12, 8))

# Line Chart
plt.subplot(2, 2, 1)

plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()

# Bar Chart
plt.subplot(2, 2, 2)

plt.bar(df["Product"], df["Sales"])
plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.grid(axis="y")

# Scatter Plot
plt.subplot(2, 2, 3)

plt.scatter(df["Quantity"], df["Total_Amount"])
plt.title("Quantity vs Total Amount")
plt.xlabel("Quantity")
plt.ylabel("Total Amount")
plt.grid()

# Pie Chart
plt.subplot(2, 2, 4)

payment_data = df.groupby("Payment_Method")["Payment_Amount"].sum()

plt.pie(payment_data.values,
        labels=payment_data.index,
        autopct="%1.1f%%")

plt.title("Payment Methods")

# Main Title
plt.suptitle("Sales Dashboard")

# Adjust spacing
plt.tight_layout()

plt.show()
```

```python
# Q25. Final Practical Challenge – Supermarket Sales Dashboard
# Read the CSV file using Pandas.
# Create:
# Monthly Sales
# Product Sales
# Payment Method Distribution
# Customer Rating Distribution
# Quantity vs Total Amount
# Use subplot(), figure size (15,10), ggplot style,
# individual titles, main title, tight_layout()
# and save as sales_dashboard.png with dpi = 300.

import pandas as pd
import matplotlib.pyplot as plt

# Use ggplot style
plt.style.use("ggplot")

# Read CSV file
df = pd.read_csv("supermarket_sales.csv")

# Create figure
plt.figure(figsize=(15, 10))

# 1. Monthly Sales - Line Chart
plt.subplot(2, 3, 1)

monthly_sales = df.groupby("Month")["Sales"].sum()

plt.plot(monthly_sales.index, monthly_sales.values, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()


# 2. Product Sales - Bar Chart
plt.subplot(2, 3, 2)

product_sales = df.groupby("Product")["Sales"].sum()

plt.bar(product_sales.index, product_sales.values)
plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.grid(axis="y")


# 3. Payment Method Distribution - Pie Chart
plt.subplot(2, 3, 3)

payment_data = df["Payment_Method"].value_counts()

plt.pie(
    payment_data.values,
    labels=payment_data.index,
    autopct="%1.1f%%"
)

plt.title("Payment Method Distribution")


# 4. Customer Rating Distribution - Histogram
plt.subplot(2, 3, 4)

plt.hist(df["Rating"], bins=5, edgecolor="black")
plt.title("Customer Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Customers")
plt.grid()


# 5. Quantity vs Total Amount - Scatter Plot
plt.subplot(2, 3, 5)

plt.scatter(df["Quantity"], df["Total_Amount"])
plt.title("Quantity vs Total Amount")
plt.xlabel("Quantity")
plt.ylabel("Total Amount")
plt.grid()


# Main Title
plt.suptitle("Supermarket Sales Dashboard")

# Adjust spacing
plt.tight_layout()

# Save dashboard
plt.savefig("sales_dashboard.png", dpi=300)

# Display dashboard
plt.show()

# Reset style
plt.style.use("default")
```
