```python
# Import Pandas

import pandas as pd

# Create a suitable student dataset

data = {
    "Name": ["Aarav", "Riya", "Sneha", "Aditya", "Priya",
             "Rahul", "Ananya", "Vihan", "Meera", "Kabir",
             "Riya", "Sneha"],
    
    "Age": [20, 21, 19, None, 20,
            22, 19, 21, None, 20,
            21, 19],
    
    "Gender": ["Male", "Female", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male",
               "Female", "Female"],
    
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai",
             "Pune", "Nashik", "Mumbai", "Pune", "Nashik",
             "Mumbai", "Pune"],
    
    "Course": ["Python", "Data Science", "Python", "AI/ML", "Data Science",
               "Python", "AI/ML", "Data Science", "Python", "AI/ML",
               "Data Science", "Python"],
    
    "Marks": [85, 92, None, 67, 88,
              76, 95, 82, 69, None,
              92, None],
    
    "Attendance": [95, 92, 88, 85, 96,
                   91, 97, 89, 94, 86,
                   92, 88],
    
    "Purchased": [1, 1, 0, 0, 1,
                  1, 1, 0, 0, 1,
                  1, 0]
}

df = pd.DataFrame(data)

# Save as CSV
df.to_csv("students.csv", index=False)

print("students.csv created successfully!")
print(df)
```

```python
# Question 1
# Replace all missing values with 0

q1_df = df.copy()

q1_df = q1_df.fillna(0)

print(q1_df)
```

```python
# Question 2
# Replace missing Marks with average Marks

q2_df = df.copy()

average_marks = q2_df["Marks"].mean()

q2_df["Marks"] = q2_df["Marks"].fillna(average_marks)

print("Average Marks:", average_marks)
print(q2_df)
```

```python
# Question 3
# Replace missing Age with average Age

q3_df = df.copy()

average_age = q3_df["Age"].mean()

q3_df["Age"] = q3_df["Age"].fillna(average_age)

print("Average Age:", average_age)
print(q3_df)
```

```python
# Question 4
# Remove rows containing missing values

q4_df = df.copy()

q4_df = q4_df.dropna()

print(q4_df)
```

```python
# Question 5
# Find duplicate records

duplicates = df[df.duplicated()]

print(duplicates)
```

```python
# Question 6
# Remove duplicate records

q6_df = df.copy()

q6_df = q6_df.drop_duplicates()

print(q6_df)
```

```python
# Question 7
# Students whose Marks are greater than average

average_marks = df["Marks"].mean()

result = df[df["Marks"] > average_marks]

print("Average Marks:", average_marks)
print(result)
```

```python
# Question 8
# Students whose Attendance is greater than 90%

result = df[df["Attendance"] > 90]

print(result)
```

```python
# Question 9
# Students whose Marks are between 70 and 90

result = df[df["Marks"].between(70, 90)]

print(result)
```

```python
# Question 10
# Female students from Pune

result = df[(df["Gender"] == "Female") & (df["City"] == "Pune")]

print(result)
```

```python
# Question 11
# Male students with Marks greater than 80

result = df[(df["Gender"] == "Male") & (df["Marks"] > 80)]

print(result)
```

```python
# Question 12
# Maximum Marks

print("Maximum Marks:", df["Marks"].max())
```

```python
# Question 13
# Minimum Marks

print("Minimum Marks:", df["Marks"].min())
```

```python
# Question 14
# Average Marks

print("Average Marks:", df["Marks"].mean())
```

```python
# Question 15
# Total Marks

print("Total Marks:", df["Marks"].sum())
```

```python
# Question 16
# Unique Course values

print(df["Course"].unique())
```

```python
# Question 17
# Number of students in each city

print(df["City"].value_counts())
```

```python
# Question 18
# Number of students in each course

print(df["Course"].value_counts())
```

```python
# Question 19
# Average Marks by City

result = df.groupby("City")["Marks"].mean()

print(result)
```

```python
# Question 20
# Course-wise aggregation

result = df.groupby("Course").agg(
    Total_Marks=("Marks", "sum"),
    Average_Marks=("Marks", "mean"),
    Minimum_Marks=("Marks", "min"),
    Maximum_Marks=("Marks", "max"),
    Total_Students=("Name", "count")
)

print(result)
```

```python
# Question 21
# Correlation between numerical columns

print(df.corr(numeric_only=True))
```

```python
# Question 22
# Manual Encoding
# Male = 1
# Female = 0

q22_df = df.copy()

q22_df["Gender_Encoded"] = q22_df["Gender"].map({
    "Male": 1,
    "Female": 0
})

print(q22_df[["Gender", "Gender_Encoded"]])
```

```python
# Question 23
# One-Hot Encoding for City and Course

q23_df = pd.get_dummies(
    df,
    columns=["City", "Course"],
    dtype=int
)

print(q23_df)
```

```python
# Question 24
# Create Features (X) and Target (y)

X = df[["Age", "Attendance", "Marks"]]

y = df["Purchased"]

print("Features (X):")
print(X)

print("\nTarget (y):")
print(y)
```

```python
# Question 25
# Complete Data Analysis

# 1. Read the CSV file
df = pd.read_csv("students.csv")

print("========== DATASET ==========")
print(df)

# 2. First 5 rows
print("\n========== FIRST 5 ROWS ==========")
print(df.head())

# 3. Last 5 rows
print("\n========== LAST 5 ROWS ==========")
print(df.tail())

# 4. Dataset information
print("\n========== DATASET INFORMATION ==========")
df.info()

# 5. Statistical summary
print("\n========== STATISTICAL SUMMARY ==========")
print(df.describe())

# 6. Check missing values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# 7. Handle missing values
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["Age"] = df["Age"].fillna(df["Age"].mean())

print("\n========== AFTER HANDLING MISSING VALUES ==========")
print(df)

# 8. Find duplicate records
print("\n========== DUPLICATE RECORDS ==========")
print(df[df.duplicated()])

# 9. Remove duplicate records
df = df.drop_duplicates()

print("\n========== AFTER REMOVING DUPLICATES ==========")
print(df)

# 10. Filter records
print("\n========== MARKS GREATER THAN 80 ==========")
print(df[df["Marks"] > 80])

print("\n========== ATTENDANCE GREATER THAN 90 ==========")
print(df[df["Attendance"] > 90])

# 11. Sort dataset
print("\n========== SORTED BY MARKS ==========")
print(df.sort_values("Marks", ascending=False))

# 12. Unique values
print("\n========== UNIQUE CITIES ==========")
print(df["City"].unique())

print("\n========== UNIQUE COURSES ==========")
print(df["Course"].unique())

# 13. value_counts()
print("\n========== CITY COUNT ==========")
print(df["City"].value_counts())

print("\n========== COURSE COUNT ==========")
print(df["Course"].value_counts())

# 14. GroupBy analysis
print("\n========== AVERAGE MARKS BY CITY ==========")
print(df.groupby("City")["Marks"].mean())

print("\n========== AVERAGE MARKS BY COURSE ==========")
print(df.groupby("Course")["Marks"].mean())

# 15. Aggregation functions
print("\n========== COURSE-WISE AGGREGATION ==========")

course_analysis = df.groupby("Course")["Marks"].agg([
    "sum",
    "mean",
    "min",
    "max",
    "count"
])

print(course_analysis)

# 16. Correlation
print("\n========== CORRELATION ==========")
print(df.corr(numeric_only=True))

# 17. Manual Encoding
df["Gender_Encoded"] = df["Gender"].map({
    "Male": 1,
    "Female": 0
})

print("\n========== MANUAL ENCODING ==========")
print(df[["Gender", "Gender_Encoded"]])

# 18. One-Hot Encoding
encoded_df = pd.get_dummies(
    df,
    columns=["City", "Course"],
    dtype=int
)

print("\n========== ONE-HOT ENCODING ==========")
print(encoded_df)

# 19. Features and Target
X = df[["Age", "Attendance", "Marks"]]
y = df["Purchased"]

print("\n========== FEATURES (X) ==========")
print(X)

print("\n========== TARGET (y) ==========")
print(y)

# 20. Save cleaned dataset
df.to_csv("cleaned_dataset.csv", index=False)

print("\ncleaned_dataset.csv saved successfully!")
```
